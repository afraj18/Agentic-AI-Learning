```python
from dotenv import load_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI

load_dotenv()
```

```python
llm = ChatGoogleGenerativeAI(model="gemma-4-31b-it")
```

```python
msg = "How many moons does earth has?"
response = llm.invoke(msg)
```

```python
print(response.text)
```

```python
response1 = llm.invoke([
    ["system","You are helpful assistant answers in short one line"],
    ["human","How many moons does earth has?"]
])
```

```python
print(response1.text)
```

```python
print(response1.text)
```

#Temperature

```python
prompt = "Write a single tagline for a quirky coffee shop, only output the tagline, nothing else"

llm = ChatGoogleGenerativeAI(model="gemini-3.8-flash",temperature=0)

resp = llm.invoke(prompt)
print(resp.text)
```

```python
llm = ChatGoogleGenerativeAI(model="gemini-3.8-flash",temperature=1)

resp = llm.invoke(prompt)
print(resp.text)
```
