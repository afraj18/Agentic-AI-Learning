```python
import chromadb
```

```python
client = chromadb.Client()
```

```python
collection = client.create_collection("news_articles_new")
```

#### Add data to db

```python
collection.add(
    ids=["id1", "id2", "id3", "id4"],
    documents=[
        "Apple is leading in a smart phone game with iPhone sales up by 35%",
        "Tesla booked a minor profit of 1 billion $ in Q2",
        "Apples are high in fiber, vitamin C, and various antioxidants",
        "SpaceX got NASA contract worth 10 billion $",
    ]
)
```

#### Get data

```python
collection.get()
```

```python
data = collection.get(include=["documents","embeddings","metadatas"])
for i, emb in enumerate(data["embeddings"]):
    print(f"documents {i+1}")
    print(data["documents"][i])
    print(f"Emb Len: {len(data["embeddings"][i])}")
    print(f"Embeddings: {data["embeddings"][i][:10]}")
    print(f"Meta Data : {data["metadatas"][i]}")
    print("\n")
```

```python
result = collection.query(
    query_texts = ["elon musk"],
    n_results = 2
)
```

```python
print(result)
```
