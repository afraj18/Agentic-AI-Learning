```python
import os
from groq import Groq

client = Groq(api_key=os.environ["GROQ_API_KEY"])
models = client.models.list()

available = sorted(
    (m for m in models.data if getattr(m, "active", True) is not False),
    key=lambda m: m.id,
)

print(f"{len(available)} active models available to your key:")
for model in available:
    print(model.id)
```
